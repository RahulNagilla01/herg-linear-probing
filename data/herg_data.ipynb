```python
from tdc.single_pred import Tox
import pandas as pd
import os
```

```python
data = Tox(name='hERG')
split = data.get_split()

train = split['train']
valid = split['valid']
test  = split['test']
```

```python
print("Train shape:", train.shape)
print("Columns:", train.columns.tolist())
print(train.head(3))
print("\nLabel distribution (train):")
print(train['Y'].value_counts())
```

```python
os.makedirs('data/raw', exist_ok=True)
train.to_csv('data/raw/herg_train.csv', index=False)
valid.to_csv('data/raw/herg_valid.csv', index=False)
test.to_csv('data/raw/herg_test.csv', index=False)
```

```python
def smiles_to_prompt(smiles, label=None):
    return f"The molecular structure represented by SMILES {smiles} was tested for hERG channel blocking activity."

train['text'] = train['Drug'].apply(smiles_to_prompt)
train[['Drug_ID', 'text', 'Y']].to_csv('data/raw/herg_train_text.csv', index=False)
print(train['text'].iloc[0])
```
